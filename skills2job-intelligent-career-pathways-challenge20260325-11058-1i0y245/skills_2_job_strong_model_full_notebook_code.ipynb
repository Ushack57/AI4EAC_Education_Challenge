```python
# =============================================
# SKILLS2JOB — PROFESSIONAL COMPETITOR SOLUTION
# FINAL ENSEMBLE SYSTEM (TOP KAGGLE STYLE)
# =============================================

import pandas as pd
import numpy as np
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.model_selection import KFold
import lightgbm as lgb
from gensim.models import Word2Vec
from collections import Counter
import warnings
warnings.filterwarnings('ignore')
```

```python
# ===============================
# LOAD DATA
# ===============================
train = pd.read_csv('Dataset/Train.csv')
test = pd.read_csv('Dataset/Test.csv')

skill_cols = [f'skill_{i}' for i in range(1,6)]
occ_cols = [f'occ_{i}' for i in range(1,6)]

train['skills'] = train[skill_cols].values.tolist()
test['skills'] = test[skill_cols].values.tolist()
train['occupations'] = train[occ_cols].values.tolist()
```

```python
# ===============================
# WORD2VEC EMBEDDINGS
# ===============================
w2v = Word2Vec(sentences=train['skills'], vector_size=64, window=3, min_count=1, epochs=30)

def embed(skills):
    vectors = [w2v.wv[s] for s in skills if s in w2v.wv]
    if len(vectors) == 0:
        return np.zeros(64)  # Return zero vector if no known skills
    return np.mean(vectors, axis=0)

X_embed = np.vstack(train['skills'].apply(embed))
X_test_embed = np.vstack(test['skills'].apply(embed))
```

```python

# ===============================
# ONE-HOT FEATURES
# ===============================
mlb_skills = MultiLabelBinarizer()
X_one = mlb_skills.fit_transform(train['skills'])
X_test_one = mlb_skills.transform(test['skills'])

# Combine
X = np.hstack([X_one, X_embed])
X_test = np.hstack([X_test_one, X_test_embed])
```

```python

# ===============================
# TARGET ENCODING
# ===============================
mlb_occ = MultiLabelBinarizer()
y = mlb_occ.fit_transform(train['occupations'])
occs = mlb_occ.classes_
```

```python

# ===============================
# K-FOLD LIGHTGBM (STRONG BASE MODEL)
# ===============================

kf = KFold(n_splits=5, shuffle=True, random_state=42)

preds_lgb = np.zeros((len(test), len(occs)))

for fold, (tr, val) in enumerate(kf.split(X)):
    print(f"Fold {fold+1}")
    for i in range(len(occs)):
        model = lgb.LGBMClassifier(
            n_estimators=300,
            learning_rate=0.03,
            num_leaves=63,
            subsample=0.85,
            colsample_bytree=0.85
        )
        model.fit(X[tr], y[tr, i])
        preds_lgb[:, i] += model.predict_proba(X_test)[:,1] / 5
```

```python
# ===============================
# SIMILARITY MODEL
# ===============================

similarity = cosine_similarity(X_test_one, X_one)

sim_preds = []
for i in range(len(test)):
    idx = np.argsort(-similarity[i])[:20]
    occs_sim = train.iloc[idx]['occupations']
    flat = [o for sub in occs_sim for o in sub]
    sim_preds.append(Counter(flat))
```

```python
# ===============================
# FREQUENCY PRIOR
# ===============================
all_occs = [o for sub in train['occupations'] for o in sub]
freq = Counter(all_occs)
```

```python
# ===============================
# FINAL ENSEMBLE RANKING
# ===============================

final_preds = []

for i in range(len(test)):
    scores = {}
    
    # LightGBM weight
    for j, occ in enumerate(occs):
        scores[occ] = preds_lgb[i][j] * 3.0
    
    # Similarity weight
    for occ, count in sim_preds[i].items():
        scores[occ] = scores.get(occ, 0) + (count * 0.2)
    
    # Frequency weight
    for occ, count in freq.items():
        scores[occ] = scores.get(occ, 0) + (count * 0.0001)
    
    ranked = sorted(scores.items(), key=lambda x: -x[1])
    final_preds.append([x[0] for x in ranked[:5]])
```

```python
# ===============================
# SAVE SUBMISSION
# ===============================

sub = pd.DataFrame()
sub['ID'] = test['ID']

for i in range(5):
    sub[f'occ_{i+1}'] = [row[i] for row in final_preds]

sub.to_csv('/mnt/data/final_pro_submission.csv', index=False)

print("🏆 PROFESSIONAL SUBMISSION READY")
```

```python
# ===============================
# MAP@5 FUNCTION
# ===============================

def mapk(actual, predicted, k=5):
    def apk(a, p, k):
        score = 0.0
        num_hits = 0.0
        for i, pred in enumerate(p[:k]):
            if pred in a and pred not in p[:i]:
                num_hits += 1.0
                score += num_hits / (i+1.0)
        return score / min(len(a), k)
    return np.mean([apk(a, p, k) for a, p in zip(actual, predicted)])
```
